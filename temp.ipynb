```python
import pandas as pd
import numpy as np
```

```python
ean = pd.read_csv('Fashione EAN 05-10-26.csv', dtype=str)
ean = ean[['PLU', 'Style', 'Colour', 'Size', 'EAN']]
ean.head()
```

```python
loc = pd.read_csv('eeb210aa-b068-4f28-8dbc-55eadb641663.csv', dtype=str)
loc['ImageURL'] = loc['ThumbnailUrl']
loc['PLU'] = loc['SKU']
loc['Pick Location'] = loc['WarehouseLocation']
loc = loc[['PLU', 'Pick Location', 'ImageURL']]

ean = pd.merge(ean, loc, on='PLU', how='left')

loc
```

```python
wr = pd.read_csv('Fonz Product List 07-10-26.csv', skiprows=1, dtype=str)
wr = wr[['PLU', 'FashBCode']]

ean = pd.merge(ean, wr, on='PLU', how='left')

ean.head()
```

```python
ean['FashBCode'] = ean['FashBCode'].fillna('-')
ean['EAN'] = ean['EAN'].fillna('-')
```

```python
ean.to_csv('Return Inventory File.csv', index=False)
```
