```python
import pandas as pd
import numpy as np
```

```python
ean = pd.read_csv('Fashione EAN 27-10-25.csv', dtype=str)
ean = ean[['PLU', 'Style', 'Colour', 'Size', 'EAN']]
ean.head()
```

```python
loc = pd.read_csv('vs_debenhams_offers_upload_export_20251029-174854_leather-company.csv', dtype=str)
temp = loc['Parent Product Images'].str.split('\n', expand=True)
loc['ImageURL'] = temp[0]
loc['PLU'] = loc['Child Reference']
loc = loc[['PLU', 'Pick Location', 'ImageURL']]

ean = pd.merge(ean, loc, on='PLU', how='left')
```

```python
wr = pd.read_excel('Fonz Product list export.xlsx', skiprows=1, dtype=str)
wr = wr[['PLU', 'FashBCode']]

ean = pd.merge(ean, wr, on='PLU', how='left')

ean.head()
```

```python
ean['FashBCode'] = ean['FashBCode'].fillna('-')
ean['EAN'] = ean['EAN'].fillna('-')
```

```python
ean.to_csv('Return Inventory File.csv', index=False)
```
